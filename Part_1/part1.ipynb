* Question 7:Vùng có tổng doanh thu cao nhất

```python
orders_p1 = orders.copy()
customers_p1 = customers.copy()
geography_p1 = geography.copy()
order_items_p1 = order_items.copy()
```

```python
#1 tổng doanh thu trên mỗi dòng đơn hàng
order_items_p1['line_revenue'] = order_items_p1['quantity'] * order_items_p1['unit_price']
order_items_p1.tail(10)
```

```python
items_per_order = order_items.groupby('order_id').size()

items_per_order.value_counts()
```

```python
multi_orders = items_per_order[items_per_order > 1]

print(multi_orders.head())
```

```python
order_revenue = order_items_p1.groupby('order_id')['line_revenue'].sum().reset_index()
order_revenue[order_revenue['order_id'] == 8]
```

```python
# Bước 3: Kết nối Đơn hàng -> Khách hàng -> Địa lý
# Đầu tiên kết nối doanh thu với thông tin đơn hàng để lấy zip khách hàng
temp_merged = pd.merge(order_revenue, orders_p1[['order_id', 'zip']], on='order_id', how='inner')
final_merged = pd.merge(temp_merged, geography_p1[['zip', 'region']], on='zip', how='inner')



region_revenue = final_merged.groupby('region')['line_revenue'].sum()
best_region = region_revenue.idxmax()
highest_revenue = region_revenue.max()

print("Tổng doanh thu theo từng vùng:")
print(region_revenue)
print(f"\nVùng tạo ra tổng doanh thu cao nhất là: {best_region}")
```

* Question 9: kích thước nào có tỷ lệ trả hàng cao
nhất

```python
order_items_p1 = order_items.copy()
returns_p1 = returns.copy()
products_p1 = products.copy()
```

```python
#size trong order
items_with_size = pd.merge(order_items_p1, products_p1[['product_id', 'size']], on='product_id', how='inner')
size_order_counts = items_with_size['size'].value_counts()

size_order_counts
```

```python
# size trong return
returns_with_size = pd.merge(returns_p1, products_p1[['product_id', 'size']], on='product_id', how='inner')
size_return_counts = returns_with_size['size'].value_counts()

size_return_counts
```

```python
return_rate_by_size = size_return_counts / size_order_counts
print(return_rate_by_size.sort_values(ascending=False))

print(f"\nKích thước có tỷ lệ trả hàng cao nhất là: {return_rate_by_size.idxmax()}")
```
